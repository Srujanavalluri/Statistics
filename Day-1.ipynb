### Scaler, Vector, Matrix and Tensor

```python
### NumPy

# Definition: NumPy (Numerical Python) is a Python library used to perform numerical calculations and work with arrays.

# Importing NumPy:
```

```python
import numpy as np
```

### Scalar (0D Array)

```python
# Definition: A scalar is a single numerical value. In NumPy, a scalar array has 0 dimensions.
```

```python
# Scalar : 0D array
import numpy as np

salary = np.array(500000)

print(salary)
print(salary.ndim)
print(type(salary))
```

### Vector (1D Array)

```python
# Definition: A vector is a one-dimensional array that stores multiple values in a single sequence.
```

```python
# Vector: 1D array
expense = np.array([6000, 4000, 10000000, 2000])

print(expense)
print(expense.ndim)
print(type(expense))
print(expense.shape)
```

### Matrix (2D Array)

```python
# Definition: A matrix is a two-dimensional array that stores values in rows and columns.
```

```python
# Matrix: 2D array
dataframe = np.array([[6000, 4000, 10000, 2000],
                      [20000, 400, 29999, 40000]])

print(dataframe)
print(dataframe.ndim)
print(type(dataframe))
print(dataframe.shape)
```

### Tensor (3D or Higher-Dimensional Array)

```python
# Definition: A tensor is a data structure that can store values in three or more dimensions. In NumPy, arrays can have any number of dimensions.
```

```python
# Tensor: 3D or more-dimensional array
array = np.array([[[1, 2, 3, 4, 5],
                   [2, 3, 1, 3, 4]],
                  [[4, 8, 5, 3, 6],
                   [2, 5, 3, 3, 2]]])

print(array)
print(array.ndim)
print(type(array))
print(array.shape)
```

### 4D Array (Four-Dimensional Array)

```python
# Definition: A 4D array is a NumPy array that contains data organized across four dimensions.
```

```python
# 4D array
array = np.array([[[[1, 2, 3, 4, 5],
                    [2, 3, 1, 3, 4]],
                   [[4, 8, 5, 3, 6],
                    [2, 5, 3, 3, 2]]],
                  [[[1, 2, 3, 4, 5],
                    [2, 3, 1, 3, 4]],
                   [[4, 8, 5, 3, 6],
                    [2, 5, 3, 3, 2]]]])

print(array)
print(array.ndim)
print(type(array))
print(array.shape)
```

## Matrix Operations

```python
# Matrix addition and subtraction
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[4, 7],
              [6, 8]])

print(A - B)
print(A + B)
```

##### Scalar Multiplication of a Matrix

```python
# Scalar multiplication
scalar = 10

A = np.array([[1, 2],
              [3, 4]])

print(scalar * A)
```

```python
# Dot product of vectors
a = np.array([1, 2])
b = np.array([3, 4])

print(a @ b)
print(np.dot(a, b))
```

```python
# Dot product of matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[4, 7],
              [6, 8]])

print(A @ B)
```

```python
# Transpose of a matrix (rows to columns and columns to rows)
A = np.array([[1, 2],
              [3, 4]])

print(np.transpose(A))
print(A.T)
```

```python
# Determinant of a matrix
A = np.array([[1, 2],
              [3, 4]])

print(np.linalg.det(A))
```

```python
# Inverse of a matrix
A = np.array([[1, 2],
              [3, 4]])

print(np.linalg.inv(A))
```
